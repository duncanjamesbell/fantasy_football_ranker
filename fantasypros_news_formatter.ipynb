```python
import pandas as pd
```

```python
news = pd.read_csv('2025 news for formatting.csv',encoding='Latin')
news
```

```python
players=[]
headlines=[]
dates=[]
fantasy_impacts=[]
categories=[]
player_news_groups = news['Group ID'].drop_duplicates().values

for g in player_news_groups:
    news_df = news[news['Group ID']==g]
    player = ''
    headline = ''
    date =''
    fantasy_impact = ''
    category = ''
    for index, row in news_df.iterrows():
        if row['Type'] == 'Player name':
            player = row['content']
        elif row['Type'] == 'Headline / Summary':
            if len(row['content']) > len(headline):
                headline = row['content']
        elif row['Type'] == 'Date':
            date = row['content']
        elif row['Type'] == 'Fantasy Impact':
            fantasy_impact = row['content']
        elif row['Type'] == 'Category':
            category = row['content']
    players.append(player)
    headlines.append(headline)
    dates.append(date)
    fantasy_impacts.append(fantasy_impact)
    categories.append(category)

formatted_news_df = pd.DataFrame()
formatted_news_df['player'] = players
formatted_news_df['date'] = dates
formatted_news_df['headline'] = headlines
formatted_news_df['fantasy_impact'] = fantasy_impacts
formatted_news_df['category'] = categories

formatted_news_df = formatted_news_df.drop_duplicates().reset_index(drop=True)
formatted_news_df
```

```python
formatted_news_df.to_csv('2025_news_formatted.csv',index=False)
```
